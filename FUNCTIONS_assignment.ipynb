```python
#4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, int(number%2) + 1):
        if number % i == 0:
            return False
    return True
```

```python
is_prime(13)
```
